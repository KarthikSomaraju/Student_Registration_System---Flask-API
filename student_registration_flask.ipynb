## 3.app.py - Main Flask Application

```python
code = """from flask import Flask, render_template, request

app = Flask(__name__)

# list to store all the students
students = []


# Homepage
@app.route("/")
def home():
    return render_template("index.html", total=len(students))


# Registration page (GET shows the form, POST takes the data)
@app.route("/register", methods=["GET", "POST"])
def register():
    if request.method == "POST":
        name = request.form["name"].strip()
        roll = request.form["roll"].strip()
        email = request.form["email"].strip()
        course = request.form["course"].strip()

        # checking the inputs
        if name == "" or roll == "" or email == "" or course == "":
            return render_template("register.html", error="Please fill all the fields!")

        if not name.replace(" ", "").isalpha():
            return render_template("register.html", error="Name should have only letters.")

        if not roll.isdigit():
            return render_template("register.html", error="Roll number should have only numbers.")

        if "@" not in email or "." not in email:
            return render_template("register.html", error="Please enter a valid email.")

        # roll number should not repeat
        for s in students:
            if s["roll"] == roll:
                return render_template("register.html", error="This roll number is already registered!")

        student = {"name": name, "roll": roll, "email": email, "course": course}
        students.append(student)
        return render_template("result.html", student=student)

    return render_template("register.html", error=None)


# Page to show all students
@app.route("/students")
def show_students():
    return render_template("students.html", students=students)


if __name__ == "__main__":
    app.run(debug=True)
"""

f = open("app.py", "w")
f.write(code)
f.close()
print("app.py created")
```
